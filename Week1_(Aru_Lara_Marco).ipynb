```python
import numpy as np
import pandas as pd


def aggregate_hourly(df):
    """Hourly means for the DataFrame whose ISD fields have had commas stripped.
    Placeholders (all-9 values) and NaN are excluded from both sum and count."""
    d = df.copy()
    d["DATE"] = pd.to_datetime(d["DATE"])

    def as_num(col):
        return pd.to_numeric(d[col], errors="coerce")

    def no_placeholder(s):
        """NaN where the value is all 9s (99, 9999, 99999, ...)."""
        digits = s.abs().astype("Int64").astype("string")
        return s.mask(digits.str.fullmatch(r"9+").fillna(False))

    def drop_quality(col):
        """TMP '+0056,1' became 561: the last digit is the QC code. Strip it, keep the sign."""
        x = as_num(col)
        return np.sign(x) * (x.abs() // 10)

    out = pd.DataFrame({"DATE": d["DATE"]})

    out["wind_speed_ms"] = no_placeholder(as_num("Wind Speed")) / 10
    out["temp_c"] = no_placeholder(drop_quality("TMP")) / 10
    out["dewpoint_c"] = no_placeholder(drop_quality("DEW")) / 10
    out["slp_hpa"] = no_placeholder(drop_quality("SLP")) / 10
    out["gust_ms"] = no_placeholder(drop_quality("OC1")) / 10

    # AA1 '01,0000,9,5' became 1000095: period | depth(4) | condition(1) | quality(1)
    aa = as_num("AA1")
    period = aa // 1_000_000
    depth = no_placeholder((aa // 100) % 10_000)
    out["precip_1h_mm"] = (depth / 10).where(period == 1)  # 1-hour totals only

    cig = no_placeholder(as_num("Cloud Height"))
    out["ceiling_unlimited_frac"] = (cig == 22000).astype(float).where(cig.notna())
    out["ceiling_m"] = cig.where(cig != 22000)

    out["visibility_m"] = no_placeholder(as_num("Visibility Distance"))

    grouped = out.set_index("DATE").resample("h")
    hourly = grouped.mean()            # mean() skips NaN
    hourly["n_reports"] = grouped.size()
    return hourly.round(3)

```

```python
hourly_weather = aggregate_hourly(weather_cleaned)
hourly_weather.head()
```
